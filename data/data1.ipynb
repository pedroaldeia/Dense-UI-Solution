# Data visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Import the CSV file into a DataFrame
data1 = pd.read_csv('data1.csv')
data2 = pd.read_csv("data2.csv")
```

```python
plt.boxplot(data1['Final'], vert=False)
plt.title('Boxplot of Final Scores')
plt.xlabel('Final')
plt.show()

plt.boxplot(data2['Final'], vert=False)
plt.title('Boxplot of Final Scores (Data2)')
plt.xlabel('Final')
plt.show()
```
